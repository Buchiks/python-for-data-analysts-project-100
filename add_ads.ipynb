```python
import pandas as pd
```

```python
ads = pd.read_csv("./ads.csv")
conversion = pd.read_json("./conversion.json")
```

```python
ads["date"] = ads["date"].str.slice(0,10)
ads = ads.rename(columns = {"date": "date_group"})
purified_conv = conversion.drop(["platform", "conversion"], axis=1)
```

```python
United = pd.merge(purified_conv, ads, on=["date_group"], how="outer")
United = United.groupby("date_group", as_index=False).agg("sum")
United = United[United["date_group"] <= "2023-08-31"]
United = United.drop(["utm_source", "utm_medium"], axis=1)
United["utm_campaign"] = United["utm_campaign"].replace(0, "none")
```

```python
United.to_json("./converison_and_ads.json")
```
